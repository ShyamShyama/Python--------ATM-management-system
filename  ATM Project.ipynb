```python
#### =======================================================================================================================================================
# function define karo :-
#=======================================================================================================================================================

def deposit(Acc_no):
    try:
        amt = int(input("Enter your amount to deposit 💰: "))
    except ValueError:
        print("❌ Invalid Input! Enter Numbers Only.")
        return
        
    if amt == 0 or amt < 0 :
        print("Invalid amount ❌")
    else :
        Acc_details[Acc_no]["Balance"] += amt
        print(f"₹{amt} successfully deposited in your account ✅")
        Acc_details[Acc_no]["History"].append(f"₹{amt} deposited 💸")
        
def balance(Acc_no):
    print(f"Balance = ₹{Acc_details[Acc_no]["Balance"]}")

def withdraw(Acc_no):
    try:
        amt = int(input("Enter your amount to withdraw 💰: "))
    except ValueError:
        print("❌ Invalid Input! Enter Numbers Only.")
        return 
        
    if amt <= 0 :
        print("Invalid amount ❌")
    elif amt > Acc_details[Acc_no]["Balance"] :
        print("⚠️ Insufficient balance")
    else:
        Acc_details[Acc_no]["Balance"] -= amt
        print(f"₹{amt} successfully withdrawn from your account ✅")
        Acc_details[Acc_no]["History"].append(f"₹{amt} withdrawn 💸")

def show_Acc_details(Acc_no) :
    print("🏬=====Account details=====🏬")
    print(f"Account No.         :  {Acc_no}")
    print(f"Account holder name :  {Acc_details[Acc_no]["Acc_holder_name"]}")
    print(f"Balance             : ₹{Acc_details[Acc_no]["Balance"]}")
    print(f"Mobile number       :  {Acc_details[Acc_no]["Mobile No."]}")

def show_history(Acc_no) :
    print("🏬=======Transaction history=======🏬")
    if len(Acc_details[Acc_no]["History"]) == 0:
        print("No history yet")
    else:
        for transaction in Acc_details[Acc_no]["History"] :
            print(transaction)
    
    

    
#============================================================================================================================================================
# function for ATM menu :- 
#===========================================================================================================================================================

def ATM_MENU(Acc_no):
        
    while True :
        print("🏬========= 🏧ATM Menu🏧 ==========🏬")
        print("1. 💸 Deposit Money")
        print("2. 💵 Withdraw Money")
        print("3. 💰 Check Balance")
        print("4. 📋 Show Account Details")
        print("5. ↩️ Log Out")
        print("6. 📜 Show History")

        try:
            choice = int(input("Enter your choice(1-6) 🏬 : "))
        except ValueError:
            print("❌ Invalid Input! Enter Numbers Only.")
            continue
            
        if choice == 1 :
            deposit(Acc_no)
            
        
        elif choice == 2 :
            withdraw(Acc_no)
        
    
        elif choice == 3 :
            balance(Acc_no)

        
        elif choice == 4 :
            show_Acc_details(Acc_no)
            
        
        elif choice == 5 :
            print("🏬 Logged out successfully 🏬")
            break

        elif choice == 6 :
            show_history(Acc_no)
          
        else:
            print("🏬 Invalid choice 🏬")


#========================================================================================================================================================
# Database------>{Dictionary}  :-
#=======================================================================================================================================================

Acc_details = {
    12345 : {
        "Acc_holder_name" : "Krishna ju",
        "PIN"             : 6363,
        "Balance"         : 100000,
        "Locked"          : False,
        "Mobile No."      : 9838347491,
        "History"         : []
    },

    12346 : {
        "Acc_holder_name" : "Radha ju",
        "PIN"             : 6464,
        "Balance"         : 5000000,
        "Locked"          : False,
        "Mobile No."      : 7240068592,
        "History"         : []
    },

    12347 : {
        "Acc_holder_name" : "Lalita sakhi",
        "PIN"             : 6565,
        "Balance"         : 300000,
        "Locked"          : False,
        "Mobile No."      : 9140497144,
        "History"         : []
    },

    12348 : {
        "Acc_holder_name" : "Visakha sakhi",
        "PIN"             : 6767,
        "Balance"         : 500000,
        "Locked"          : False,
        "Mobile No."      : 9999626288,
        "History"         : []
    },

    12349 : {
        "Acc_holder_name" : "Priyavadini sakhi",
        "PIN"             : 6868,
        "Balance"         : 10000000,
        "Locked"          : False,
        "Mobile No."      : 9999988888,
        "History"         : []
    }
}

#========================================================================================================================================================
# Main program :-
#=========================================================================================================================================================

while True :
    try:
        Acc_no = int(input("Plz enter your Account Number 🏬------>(Press 0 to exit ATM) : "))
    except ValueError:
        print("❌ Invalid Input! Enter Numbers Only.")
        continue

    if Acc_no == 0 :
        print("🏬 Thank you for using our ATM. Visit Again! 🙏🏬")
        break
        
    elif Acc_no in Acc_details:
        
        if Acc_details[Acc_no]["Locked"] == True:
            print("Sorry,your account is🔒 locked.Plz contact your bank 🏬.")
            continue

        print(f"Hello {Acc_details[Acc_no]["Acc_holder_name"]}")

        count = 3
        while count > 0 :
            try:
                password = int(input("Enter your PIN 🏬 : "))
            except ValueError:
                print("❌ Invalid Input! Enter Numbers Only.🏬")
                continue
                
            if password == Acc_details[Acc_no]["PIN"]:
                print("Login Successful ✅")
                
                ATM_MENU(Acc_no)
                break
            
            else:
                print(f"Wrong PIN 🏬. Attempts left : {count-1}")
                count -= 1
                
        else:
            Acc_details[Acc_no]["Locked"] = True
            print("🏬 Account Locked 🔒")
        
    else:
        print("🔍❌ Account not found 🏬")
        continue
```
